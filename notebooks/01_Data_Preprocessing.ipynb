## Convert raw fan-controller `.txt` log to CSV

Keeps only the Time / Duty / Frequency / RPM columns and writes a cleaned CSV (`z=2_r=0_cleaned.csv`) used later in the duty-cycle sweep analysis.

```python
import csv

input_file = "z=2,r=0.txt"
output_file = "z=2_r=0_cleaned.csv"
columns_to_keep = [0, 1, 3, 4]  # zero-indexed
header = ["Time (s)", "Duty (%)", "Frequency (Hz)", "RPM"]
with open(input_file, 'r') as txt_in, open(output_file, 'w', newline='') as csv_out:
    reader = csv.reader(txt_in)
    writer = csv.writer(csv_out)
    writer.writerow(header)
    for row in reader:
        if not row or len(row) < 5:
            continue  # Skip empty or incomplete lines
        selected = [row[i].strip() for i in columns_to_keep]
        writer.writerow(selected)

print(f"✅ CSV saved as '{output_file}'")
```
