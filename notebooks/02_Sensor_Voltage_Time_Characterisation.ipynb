# 02 · Sensor Voltage–Time Characterisation

**Purpose.** Characterise the raw anemometer voltage signal over time for each probe position `(z, r)` as the fan duty cycle is stepped, then summarise the derived velocity–RPM and voltage–time relationships across all measured positions.

**Story:** raw per-position time series (with duty-change markers) → summary time-series views across all positions → derived velocity-vs-RPM views.


## Voltage vs. Time — axial reading, z = 1, r = 0

Loads the raw Excel export, detects duty-cycle change points, and overlays them on the voltage trace.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Excel file
file_path = "z=1, r=0.xlsx"
excel_data = pd.ExcelFile(file_path)
df = excel_data.parse('11')

# Find indices where duty cycle changes
duty_change_points = df[df['Duty'].diff() != 0].index.tolist()
duty_change_points.insert(0, 0)  # Ensure the first point is included

# Collect segment info: start time, duty, rpm
segments = []
for idx in duty_change_points:
    time = df['Time'].iloc[idx]
    duty = df['Duty'].iloc[idx]
    rpm = df['RPM'].iloc[idx]
    segments.append((time, duty, rpm))

# Plot Voltage vs Time
plt.figure(figsize=(12, 6))
plt.plot(df['Time'], df['Voltage'], color='blue', linewidth=1, label='Voltage')

# Add vertical lines with annotations for all duty changes
for time, duty, rpm in segments:
    plt.axvline(x=time, color='grey', linestyle='--', linewidth=1)
    plt.text(time, max(df['Voltage']) * 0.95, f'{duty}%\n{rpm} RPM',
             rotation=90, verticalalignment='top', horizontalalignment='right',
             fontsize=8, color='black')

plt.title('Voltage vs Time for z = 1, r = 0 [axial reading I]')
plt.xlabel('Time (s)')
plt.ylabel('Voltage (V)')
plt.grid(True)
plt.tight_layout()
plt.show()
```
