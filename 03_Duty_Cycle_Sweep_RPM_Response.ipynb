# 03 · Duty-Cycle Sweep RPM Response

**Purpose.** Up/down duty-cycle sweeps logged at two PWM switching frequencies (20 kHz and 25 kHz), showing RPM response and hysteresis between the up- and down-sweep directions.

**Note on duplication:** the original notebook contained earlier plain versions of both sweep plots that only added axis ticks/aesthetics in a later cell. Only the improved (final) versions are kept here; the earlier versions are preserved in the Experiments notebook.

## 20 kHz sweep — Duty Cycle vs. RPM

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv("up and down, 0.5%.csv")

# Extract up and down sweep data
duty_up = df["Duty Cycle (%)"]
rpm_up = df["RPM"]

duty_down = df["Duty cycle"]
rpm_down = df["RPM.1"]

# Plot both sweeps
plt.figure(figsize=(10, 5))
plt.plot(duty_up, rpm_up, label='Up Sweep', color='blue', marker='o')
plt.plot(duty_down, rpm_down, label='Down Sweep', color='red', marker='x')

# Axis labels and title
plt.xlabel("Duty Cycle (%)")
plt.ylabel("RPM")
plt.title("Duty Cycle vs RPM for 20kHz")

# Custom tick marks
plt.xticks(range(0, 105, 5))  # 0, 5, 10, ..., 100
plt.yticks(range(250, 5001, 250))  # 250, 500, ..., 5000

# Aesthetics
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

## 20 kHz sweep — averaged up/down RPM table

Groups the same 20 kHz sweep CSV by duty cycle and reports the mean RPM for the up- and down-sweep directions side by side.

```python
import pandas as pd

# Load the CSV
df = pd.read_csv("up and down, 0.5%.csv")

# Group and average for up sweep
up_avg = df.groupby("Duty Cycle (%)")["RPM"].mean().reset_index()
up_avg.columns = ["Duty Cycle (%)", "Up Sweep RPM"]

# Group and average for down sweep
down_avg = df.groupby("Duty cycle")["RPM.1"].mean().reset_index()
down_avg.columns = ["Duty Cycle (%)", "Down Sweep RPM"]

# Merge on Duty Cycle
combined = pd.merge(up_avg, down_avg, on="Duty Cycle (%)", how="outer").sort_values("Duty Cycle (%)")

# Optional: round for readability
combined = combined.round(2)

# Print the table
print(combined)

# Optional: save to CSV
# combined.to_csv("averaged_rpm_table.csv", index=False)
```

## 25 kHz sweep — Duty Cycle vs. RPM

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV (adjust path if needed)
df = pd.read_csv("duty_freq_rpm_data_updown_25khz.csv", header=0)

# Drop fully empty columns if any
df = df.dropna(axis=1, how='all')

# Rename columns appropriately
df.columns = ['Duty_Up', 'Freq_Up', 'RPM_Up', 'Duty_Down', 'Freq_Down', 'RPM_Down']

# Drop rows that are completely NaN
df.dropna(how='all', inplace=True)

# Plot
plt.figure(figsize=(10, 5))
plt.plot(df['Duty_Up'], df['RPM_Up'], label='Up Sweep', marker='o', color='blue')
plt.plot(df['Duty_Down'], df['RPM_Down'], label='Down Sweep', marker='x', color='red')

# Labels and title
plt.xlabel("Duty Cycle (%)")
plt.ylabel("RPM")
plt.title("Duty Cycle vs RPM for 25 kHz")

# Set custom ticks
plt.xticks(range(0, 105, 5))           # X-axis: 0 to 100 in steps of 5
plt.yticks(range(250, 5001, 250))      # Y-axis: 250 to 5000 in steps of 250

# Aesthetics
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
