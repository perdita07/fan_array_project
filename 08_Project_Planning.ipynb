# 08 · Project Planning

**Purpose.** Administrative work-plan / Gantt chart for the project timeline. Not a scientific analysis — kept separate from the experimental notebooks so the analysis notebooks stay focused on results.


```python
import matplotlib.pyplot as plt
import pandas as pd

tasks = [
    "Fan Characterisation & Control Setup",
    "Wind Profile Input",
    "Frequency-Domain Analysis",
    "Inverse Mapping Development",
    "Experimental Validation",
]

start_weeks = [0, 2, 5, 8, 12]
durations = [2, 3, 3, 4, 3]

df = pd.DataFrame({
    "Task": tasks,
    "Start": start_weeks,
    "Duration": durations
})

fig, ax = plt.subplots(figsize=(10, 6))

for i, row in df.iterrows():
    ax.barh(row["Task"], row["Duration"], left=row["Start"], height=0.5, align="center")

ax.set_xlabel("Weeks")
ax.set_title("Work Plan")
ax.invert_yaxis()
ax.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
